```python
import cv2
import os
import numpy as np
import matplotlib.pyplot as plt

dataset_path = "C:/Users/Acer/Documents/ISWAN DATAS/ORB ALGORITHM/DATASET/FINGERPRINT DATASET/dataset/clean_line"
query_path = "C:/Users/Acer/Documents/ISWAN DATAS/ORB ALGORITHM/DATASET/FINGERPRINT DATASET/dataset/data_test/1__M_Right_middle_finger.BMP"

orb = cv2.ORB_create(nfeatures=2000)
bf = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True)



query_img = cv2.imread(query_path, 0)
kp_q, des_q = orb.detectAndCompute(query_img, None)



best_score = 0
best_match = None

for root, dirs, files in os.walk(dataset_path):
    for file in files:
        if file.lower().endswith(".bmp"):
            
            path = os.path.join(root, file)
            img = cv2.imread(path, 0)

            kp, des = orb.detectAndCompute(img, None)

            if des is None or des_q is None:
                continue

            matches = bf.match(des_q, des)
            matches = sorted(matches, key=lambda x: x.distance)

            good = [m for m in matches if m.distance < 50]

            score = len(good)

            if score > best_score:
                best_score = score
                best_match = path
                
        
print("Best Match:", best_match)
print("Score:", best_score)

if best_score > 30:
    print("Sidik jari COCOK")
else:
    print("Sidik jari TIDAK cocok")
    
    

# Load gambar best match
best_img = cv2.imread(best_match, 0)

# Resize biar sama ukuran (WAJIB)
h = 300
w = 300

query_resized = cv2.resize(query_img, (w, h))
best_resized = cv2.resize(best_img, (w, h))

# Gabungkan kiri-kanan
combined = np.hstack((query_resized, best_resized))

# Tampilkan
plt.imshow(combined, cmap='gray')
plt.title("Query (Kiri) vs Best Match (Kanan)")
plt.axis('off')
plt.show()
```
